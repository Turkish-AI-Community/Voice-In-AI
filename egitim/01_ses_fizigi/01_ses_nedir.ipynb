# 1. Ses Nedir? Mekanik Dalga Teorisi

## 1.1 Ses Nedir?

**Ses, bir ortam içerisindeki parçacıkların titreşimleri sonucu oluşan ve enerji taşıyan mekanik bir boyuna basınç dalgadır.**

Burada özellikle üç kavramı ayırmak gerekir.

- Ses madde değildir.
- Ses enerjidir.
- Enerjiyi taşıyan şey ise ortamdaki parçacıkların titreşimidir.

**Her sesin başlangıcında mutlaka titreşen bir kaynak bulunur.**

Bu kaynak;

- insan ses telleri olabilir,
- gitar teli olabilir,
- hoparlör membranı olabilir,
- davul derisi olabilir,
- motor olabilir,
- deprem sırasında kayan kaya kütleleri bile olabilir.

Ortak nokta şudur:

- Kaynak dengede durmaz.
- İleri-geri hareket eder.
- Bu hareket çevresindeki ortama aktarılır.

Örneğin bir diyapazonu düşünelim. Diyapazon sabit konumdayken etrafındaki hava molekülleri de olağan hareketlerine devam ederler. Ne zamanki diyapazonun kollarına vurup da onları titreştirdiğimizde kolların ileri geri hareketi ile çevrelerinde hava moleküllerine dışarıdan bir başka kuvvet daha uygulanmış olur. 

Diyapazon kollarının ileri hareketi hava moleküllerini ileri iter ve kollara hemen yakın konumdaki molekülleri birbirine yaklaştırır, sıkıştırır. Buradaki basınç artar. Bu bölgeye **Sıkışma (Compression)** bölgesi denir. 

Daha sonra kolun geri hareketi ile moleküller birbirinden uzaklaşır ve böylece basınç azalır. Bu bölgeye ise **Genleşme (Rarefaction)** denir.

Ses sürekli oluşan basınç değişimlerinden meydana gelir. İlk molekül ikinciyi iter, ikinci üçüncüyü iter, üçüncü dördüncüyü iter ve zincir bu şekilde devam eder. **Enerji bir basınç dalgası halinde yayılır.**

**İşte bu yüzden ses boşlukta duyulmaz. Sesin yayılması için parçacıklara ihtiyacı vardır.**

## 1.3 Sesin Maddesel Ortamlardaki Yayılımı

Gazlarda parçacıklar birbirinden oldukça uzaktadır. Bu nedenle iletim daha yavaş ilerler. Ses 20°C kuru havada yaklaşık 343 m/s hızda ilerler.

Sıvılarda ise moleküller gaz moleküllerine göre daha yakındır, Bu yüzden enerji aktarımı daha kolaydır. Ses daha hızlı ilerler. Suda yaklaşık 1480 m/s hızda ilerler. Balinaların kilometrelerce uzaktan haberleşebilmesinin bir nedeni de budur.

Katılarda ise atomlar kristal yapılar içinde birbirine çok daha sıkı bağlıdır. Bir atom titreştiğinde komşusunu neredeyse anında etkiler. Bu yüzden çok daha hızlı yayılır. Ses hızı katılarda 3000-6000 m/s aralığında çeşitlilik gösterirken çelikte yaklaşık 5100 m/s hızda ilerler. Bu nedenle tren rayına kulaklarınızı koyarsanız, trenin titreşimlerini havadan çok daha önce duyabilirsiniz. 


Ses hızı sıcaklığa da bağlıdır.

İdeal gaz için ses hızı:

$$
c=\sqrt{\gamma\frac{RT}{M}}
$$

Burada;

- $c$ : Ses hızı (m/s)
- $\gamma$ : Adyabatik katsayı (hava için yaklaşık **1.4**)
- $R$ : Evrensel gaz sabiti ($8.314\ \mathrm{J/(mol\cdot K)}$)
- $T$ : Mutlak sıcaklık (**Kelvin**)
- $M$ : Gazın molar kütlesi (hava için yaklaşık **0.0289652 kg/mol**)

> **Dikkat:** Bu denklemde sıcaklık **Santigrat (°C)** değil, **Kelvin (K)** cinsinden kullanılmalıdır.

Santigrat sıcaklığını Kelvin'e çevirmek için:

$$
T_K = T_{^\circ C} + 273.15
$$

Örneğin:

- $0^\circ C = 273.15\ K$
- $20^\circ C = 293.15\ K$
- $37^\circ C = 310.15\ K$

Bu dönüşüm yerine yazıldığında denklem şu hale gelir:

$$
c
=
331.3
\sqrt{1+\frac{T}{273.15}}
$$

Burada $T$, artık **°C cinsinden sıcaklığı** ifade etmektedir.

Bu denklemin düşük ve orta sıcaklıklar için doğrusal yaklaşımı ise şöyledir:

$$
c \approx 331.3 + 0.606T
$$

Bu yaklaşık formül günlük mühendislik hesaplamalarında yaygın olarak kullanılır.

```python
import numpy as np

def ses_hizi(T_celsius):
    """Sıcaklığa bağlı ses hızı (m/s)"""
    return 331.3 * np.sqrt(1 + T_celsius / 273.15)

for T in [0, 20, 37]:
    print(f"T={T}°C  →  c = {ses_hizi(T):.1f} m/s")
# T=0°C  →  c = 331.3 m/s
# T=20°C →  c = 343.2 m/s
# T=37°C →  c = 352.9 m/s
```

Sıcaklık arttıkça ses hızı da artar çünkü yüksek sıcaklıklarda hava molekülleri de daha hızl hareket ederler. 

Ayrıca nemli havada kuru havaya göre sesin daha hızlı iletilmesini sağlar. 

# 2. Temel Akustik Parametreler

## 2.1 Frekans (f)

Birim zamandaki tam titreşim sayısıdır. Birimi Hertz (Hz).

$$f = \frac{1}{T}$$

Burada $T$ periyottur (bir tam titreşimin süresi, saniye cinsinden).

| Ses Türü                          | Frekans Aralığı       |
| --------------------------------- | --------------------- |
| İnfra-ses                         | < 20 Hz               |
| **İnsan işitme aralığı**          | **20 Hz – 20.000 Hz** |
| Ultrasonik                        | > 20.000 Hz           |
| Konuşma temel frekansı (F0) erkek | 85–180 Hz             |
| Konuşma temel frekansı (F0) kadın | 165–255 Hz            |
| Telefon bandı (narrowband)        | 300–3400 Hz           |
| Geniş bant konuşma (wideband)     | 50–7000 Hz 

```python
import numpy as np
import matplotlib.pyplot as plt

# Temel bir sinüs dalgası
fs = 44100         # örnekleme hızı (Hz)
duration = 0.05    # süre (saniye)
f = 440            # La notası (A4)

t = np.linspace(0, duration, int(fs * duration), endpoint=False)
sinyal = np.sin(2 * np.pi * f * t)

plt.figure(figsize=(10, 3))
plt.plot(t * 1000, sinyal)
plt.xlabel("Zaman (ms)")
plt.ylabel("Genlik (normalize)")
plt.title(f"f={f} Hz Sinüs Dalgası — T = {1000/f:.2f} ms")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 2.2 Genlik (Amplitude)

Dalganın tepe basınç değeridir. Genlik, sesin şiddetini (loudness) belirler. Ses basınç seviyesi genellikle Pascal (Pa) cinsinden ölçülür; insan kulağının duyma eşiği yaklaşık $20 \times 10^{-6}$ Pa = 20 µPa'dır.

$$A(t) = A_0 \sin(2\pi f t + \phi)$$

## 2.3 Faz (Phase)

Dalganın referans noktasına göre ne kadar kaydırıldığını belirtir. Radyan ($\phi$) veya derece cinsinden ifade edilir.

```python
t = np.linspace(0, 0.01, 1000)
f = 440

# Farklı fazlardaki sinyaller
sinyal_0   = np.sin(2 * np.pi * f * t)
sinyal_90  = np.sin(2 * np.pi * f * t + np.pi/2)    # +90° faz
sinyal_180 = np.sin(2 * np.pi * f * t + np.pi)      # +180° faz (ters polar)

# İki sinyal toplandığında 0° + 180° = SES YOK (faz iptali)
toplam = sinyal_0 + sinyal_180
print(f"Faz iptali sonucu max genlik: {np.max(np.abs(toplam)):.6f}")
# ≈ 0.000000 (tam iptal)
```

> Pratik önem: Stereo kayıtlarda veya çok mikrofonlu sistemlerde faz uyumsuzluğu faz iptali (phase cancellation) yaratarak belirli frekansları sönümler. Özellikle yakın mikrofon tekniklerinde kritik bir sorundur.

## 2.4 Dalga Boyu (λ)

Bir tam titreşimin kapladığı uzaklık:

$$\lambda = \frac{c}{f}$$

| Frekans   | Dalga Boyu (20°C'de) |
| --------- | -------------------- |
| 20 Hz     | 17.16 m              |
| 1000 Hz   | 34.3 cm              |
| 5000 Hz   | 6.86 cm              |

## 2.5 Moleküllerin Hareketinden Dalga Grafiğine

Önceki bölümde sesin nasıl oluştuğunu gördük.

Titreşen bir kaynak, çevresindeki hava moleküllerini ileri ve geri hareket ettirerek bölgede yüksek basınç **(compression)** ve düşük basınç **(rarefaction)** alanları oluşturur.

Bu basınç değişimleri molekülden moleküle aktarılır ve ortam boyunca ilerleyen bir boyuna basınç dalgası meydana gelir.

Buraya kadar baktığımız şey tamamen fiziksel olaydı. 
Fakat mühendislikte ve yapay zekâda milyonlarca molekülü tek tek modellemek mümkün değildir.

Bunun yerine, belirli bir noktadaki basıncın zamana göre nasıl değiştiğini ölçeriz.

Örneğin mikrofonun bulunduğu noktayı düşünelim.

- Dalga mikrofona ulaştığında önce yüksek basınç gelir.
- Ardından normal basınç.
- Sonra düşük basınç.
- Sonra tekrar normal basınç.

Bu değişim sürekli tekrar eder.

İşte çizilen sinüs eğrisi aslında mikrofonun bulunduğu tek bir noktadaki basıncın zaman içinde nasıl değiştiğini gösterir.

**Frekans** için birim zamandaki titreşim sayısıdır demiştik. Bir saniyedeki compression-rarefaction (sıkışma-genleşme) döngüsünün sayısı olarak da ifade edebiliriz. 

Frekans arttıkça,

* moleküller daha hızlı ileri-geri titreşir,
* sıkışma ve seyrelme bölgeleri daha sık oluşur,
* kulağımız sesi daha ince (yüksek perde) olarak algılar.

**Genlik** ise basınç değişiminin ne kadar büyük olduğunu ifade eder.

Küçük genlikte normal basın. ile sıkışma bölgeleri arasındaki fark küçüktür. Büyük genlikte ise çok daha belirgin sıkışmalar oluşur. 

Yani büyük genlikte moleküller denge konumlarından daha fazla saparlar. Bu da daha büyük basınç değişimi oluşturur. 

Sinüs grafiğinde de bunun karşılığı tepe yüksekliğidir. 

Mikrofon örneğini ele alırsak da mkrofon diyaframına uygulanan maksimum basınç değişimidir.

**Faz**, aynı titreşim döngüsünün hangi anında bulunduğumuzu gösterir.

İki farklı hoparlör düşünelim. İkisi de 440 Hz üretiyor. Birincisi tam sıkışma oluştururken ikincisi aynı anda seyrelme oluşturuyorsa iki dalga birbirini söndürmeye başlar.

Çünkü birinin oluşturduğu yüksek basınç, diğerinin oluşturduğu düşük basıncı dengeler. Mikrofonun gördüğü toplam basınç neredeyse sıfır olur.

İşte faz iptali tam olarak budur.

Dolayısıyla faz, basınç dalgalarının birbirlerine göre zamanlamasını ifade eder.

**Dalga boyu**, bir sıkışma bölgesi ile ondan sonraki sıkışma bölgesi arasındaki fiziksel uzaklıktır.

Bu uzaklık dalganın uzayda kapladığı mesafedir.

Burada çok güzel bir ilişki ortaya çıkar:

Ses hızı 343 m/s ise ve kaynak 343 Hz titreşiyorsa her saniyede 343 sıkışma oluşur. Dalga bir saniyede 343 metre ilerlediğine göre her sıkışma arasında 1 metre bulunur.

Yani

$\lambda = \frac{c}{f}$

denklemindeki fiziksel anlam tam olarak budur.

* Ses hızı (c): Basınç dalgasının ortamda ilerleme hızı.
* Frekans (f): Kaynağın saniyede oluşturduğu sıkışma-seyrelme döngüsü sayısı.
* Dalga boyu (λ): Art arda gelen iki sıkışma (veya iki seyrelme) cephesi arasındaki mesafe.

Bu yüzden frekans arttıkça, aynı hızla ilerleyen dalga daha kısa aralıklarla sıkışmalar oluşturur ve dalga boyu küçülür. Frekans azaldığında ise sıkışmalar arasındaki mesafe artar ve dalga boyu büyür.

Özetle,
* Frekans → Moleküller ne kadar hızlı ileri-geri salınıyor?
* Genlik → Moleküller denge konumlarından ne kadar uzaklaşıyor ve ne kadar büyük basınç değişimi oluşturuyor?
* Faz → İki basınç dalgasının sıkışma ve seyrelme bölgeleri zaman açısından ne kadar hizalı?
* Dalga boyu → Uzayda iki ardışık sıkışma cephesi arasındaki fiziksel mesafe.

# 3. Desibel ve Ses Şiddeti

Ses dalgalarını anlatırken bazen basınç farkları çok küçük, bazen de çok büyük olabilir. İnsan kulağı ise yaklaşık $10^{12}$ katlık bir dinamik aralığı algılar. Bu yüzden doğrusal bir ölçek yerine, sesin büyüklüğünü ifade etmek için çoğunlukla logaritmik desibel (dB) ölçeği kullanılır.

Logaritmik ölçekler, çok küçük ve çok büyük değerleri aynı eksende daha anlamlı biçimde göstermemizi sağlar. Ses işleme, akustik ölçüm ve mikrofon kalibrasyonunda dB kullanımı bu yüzden standarttır.

## 3.1 Desibel Tanımı

Desibel, bir büyüklüğün referans bir değere göre ne kadar büyük ya da küçük olduğunu ifade eden logaritmik bir orandır. Ses alanında iki temel kullanım vardır:

* **Ses basınç seviyesi** (SPL, Sound Pressure Level)
* **Ses şiddeti seviyesi** (SIL, Sound Intensity Level)

Ses Basınç Seviyesi (SPL):

$$L_p = 20 \log_{10}\left(\frac{p}{p_0}\right) \quad \text{dB SPL}$$

Burada $p_0 = 20 \times 10^{-6}$ Pa (işitme eşiği referansı).

Ses Şiddeti Seviyesi (SIL):

$$L_I = 10 \log_{10}\left(\frac{I}{I_0}\right) \quad \text{dB SIL}$$

Burada $I_0 = 10^{-12}$ W/m² (referans yoğunluk).

### 3.1.1 Ses Basınç Seviyesi (SPL)


Ses, fiziksel olarak ortam basıncındaki çok küçük değişimlerdir. Mikrofonun veya kulağın algıladığı şey bu basınç farkıdır. Bu yüzden ses basıncı için kullanılan formül:

$$
L_p = 20 \log_{10}\left(\frac{p}{p_0}\right) \quad \text{dB SPL}
$$

Burada:

* $L_p$: ses basınç seviyesi
* $p$: ölçülen ses basıncı
* $p_0 = 20 \times 10^{-6}\ \text{Pa}$: işitme eşiği referansı

Bu referans değer, yaklaşık olarak sağlıklı bir insan kulağının 1 kHz civarında duyabildiği en düşük basınç kabul edilir.

Buradaki 20 katsayısı önemlidir. Çünkü basınç bir genlik büyüklüğüdür; güç ise genliğin karesiyle orantılıdır. dB tanımı güç için 10 log ile yazılırken, basınç gibi genlik büyüklükleri için kare ilişkisi nedeniyle 20 log kullanılır.

### 3.1.2 Ses Şiddeti Seviyesi (SIL)

Ses şiddeti, birim alandan geçen akustik güç akışını ifade eder. Yani fiziksel olarak enerji taşınımını anlatır. Formülü:

$$
L_I = 10 \log_{10}\left(\frac{I}{I_0}\right) \quad \text{dB SIL}
$$

Burada:

* $L_I$: ses şiddeti seviyesi
* $I$: ölçülen ses şiddeti
* $I_0 = 10^{-12}\ \text{W/m}^2$: referans şiddet

Bu formülde 10 katsayısı kullanılır, çünkü şiddet doğrudan bir güç/enerji büyüklüğüdür.

### 3.1.3 Neden Logaritmik Ölçek Kullanılır?

İnsan kulağı lineer değil, yaklaşık logaritmik algı yapar. Yani aynı fiziksel artış, her seviyede aynı algısal farkı yaratmaz.

Örneğin:

* 30 dB ile 40 dB arasındaki fark,
* 90 dB ile 100 dB arasındaki fark

fiziksel olarak aynı dB artışıdır, ama algısal anlamı bağlama göre farklı olabilir.

Bu yüzden dB, hem ölçüm hem de algı açısından daha kullanışlıdır.

## 3.2 Önemli dB Referans Değerleri

| Ses Kaynağı    | dB SPL  | Açıklama        |
| -------------- | ------- | --------------- |
| İşitme eşiği   | 0 dB    | $p_0$ referansı |
| Fısıltı        | ~30 dB  |                 |
| Normal konuşma | ~60 dB  | 1 m mesafeden   |
| Gürültülü ofis | ~70 dB  |                 |
| Konsert        | ~110 dB |                 |
| Jet motoru     | ~140 dB | Ağrı eşiği      |

Burada dikkat edilmesi gereken nokta şu: dB değerleri yalnızca kaynağa değil, ölçüm mesafesine, ortamın akustiğine ve yansımaya da bağlıdır.

```python
import numpy as np

def amplitude_to_db(amplitude, ref=1.0):
    """Genlik değerini dB'e çevir"""
    return 20 * np.log10(np.abs(amplitude) / ref + 1e-10)

def db_to_amplitude(db, ref=1.0):
    """dB değerini genliğe çevir"""
    return ref * 10 ** (db / 20)

def power_to_db(power, ref=1.0):
    """Güç değerini dB'e çevir"""
    return 10 * np.log10(power / ref + 1e-10)

# Örnekler
print(f"0.001 genlik  → {amplitude_to_db(0.001):.1f} dB")   # -60 dB
print(f"-20 dB        → genlik = {db_to_amplitude(-20):.4f}")  # 0.1
print(f"İki kat güç   → +{power_to_db(2):.2f} dB")            # ~+3 dB
print(f"İki kat basınç → +{amplitude_to_db(2):.2f} dB")       # ~+6 dB
```

## 3.3 dB’nin Fiziksel Anlamı

dB ölçeğinde küçük görünen farklar fiziksel olarak önemli değişimlere karşılık gelir.

**+3 dB**

Güç yaklaşık iki katına çıkar.

$$
+3 \text{ dB} \approx \times 2 \text{ güç}
$$

**+6 dB**

Basınç genliği yaklaşık iki katına çıkar.

$$
+6 \text{ dB} \approx \times 2 \text{ basınç}
$$

Çünkü genlik iki katına çıkarsa, güç yaklaşık dört katına çıkar; bu da 10 log ölçeğinde yaklaşık 6 dB eder.

**+10 dB**

Fiziksel olarak güç yaklaşık 10 kat artar. İnsan kulağı bunu çoğu durumda yaklaşık iki kat daha yüksek gibi algılayabilir. Bu, kesin bir fizik yasası değil, psikoakustik bir yaklaşımdır.

# 4. Mono, Stereo ve Çok Kanallı Ses

Gerçek dünyada bir ses kaydı, yalnızca “sesin kendisi” değildir; aynı zamanda bu sesin kaç ayrı kanaldan kaydedildiğini de içerir. Bir ses dosyasının mono, stereo veya çok kanallı olması, aslında sinyalin kaç bağımsız ses akışı taşıdığını anlatır.

Burada temel soru şudur:

Aynı fiziksel ses olayı neden farklı kanal yapılarında temsil edilebilir?

Cevap, kayıt ve oynatma sistemlerinin uzamsal bilgi taşıma ihtiyacında yatar.

## 4.1 Mono Ses

**Mono**, tek kanallı sestir.

Matematiksel olarak bir mono kayıt genellikle şu şekilde temsil edilir:

`shape -> (N,)`

Burada:

`N`, toplam örnek sayısıdır. Her örnek, tek bir zaman noktasındaki ses genliğini temsil eder.

Mono sesin özelliği şudur: Tüm ses bilgisi tek bir kanal içinde toplanır. Yani sol ve sağ ayrımı yoktur. Bu yüzden mono kayıtlar konuşma odaklı sistemlerde çok yaygındır.

Örnek kullanım alanları:

* ASR (Automatic Speech Recognition)
* Telefon görüşmeleri
* Basit ses komut sistemleri
* Tek mikrofonlu kayıtlar

Mono, özellikle konuşma tanıma için tercih edilir çünkü modelin gereksiz uzamsal bilgiyle uğraşmasını engeller. Konuşmanın içeriği önemlidir; “hangi hoparlörden geldiği” çoğu zaman ikincil bilgidir.

## 4.2 Stereo Ses

**Stereo**, iki kanallı sestir.

Tipik gösterim:

shape -> (N, 2)

Burada:

* ilk kanal genellikle sol kanaldır,
* ikinci kanal genellikle sağ kanaldır.

Stereo sesin temel amacı, ses sahnesinde bir uzamsal genişlik hissi oluşturmaktır. Örneğin müzikte enstrümanlar sola-sağa dağıtılarak daha doğal bir dinleme deneyimi sağlanır.

Stereo’da her kanal aynı sinyali taşımaz. 

Bazen:

* sol ve sağ kanalda farklı enstrümanlar,
* farklı mikrofon konumları,
* farklı işlenmiş sinyaller

bulunabilir.

Bu nedenle stereo kayıt, tek bir dalga değil; zamana göre değişen iki ayrı dalga dizisi gibi düşünülmelidir.

## 4.3 Çok Kanallı Ses

Stereo’nun ötesinde daha fazla kanal varsa buna çok kanallı ses denir.

Örnekler:

* 2.1 → 3 kanal
* 5.1 → 6 kanal
* 7.1 → 8 kanal
* Ambisonics → 4 ve üzeri kanal

Buradaki amaç, yalnızca sol-sağ ayrımı değil, aynı zamanda çevresel konum hissi oluşturmaktır.

Örneğin:

* 5.1 surround sisteminde ön, arka, merkez ve düşük frekans kanalları ayrı tutulur.
* Ambisonics ise 360° ve VR ses uygulamalarında, sesin yön bilgisiyle birlikte temsil edilmesini sağlar.

Çok kanallı kayıtlar özellikle şu alanlarda önemlidir:

* sinema
* oyun ses tasarımı
* sanal gerçeklik
* sahne kayıtları
* çok mikrofonlu prodüksiyonlar

```python
import soundfile as sf
import numpy as np

# Mono ses: shape → (N,)
# Stereo ses: shape → (N, 2)
# 5.1 ses: shape → (N, 6)

# Stereo oluşturma
fs = 44100
duration = 1.0
t = np.linspace(0, duration, int(fs * duration))

sol_kanal   = np.sin(2 * np.pi * 440 * t)   # Sol: La
sag_kanal   = np.sin(2 * np.pi * 880 * t)   # Sağ: La oktav üstü

stereo = np.stack([sol_kanal, sag_kanal], axis=1)  # (N, 2)
print(f"Stereo shape: {stereo.shape}")

# Stereo'yu mono'ya dönüştürme (ortalama)
mono = np.mean(stereo, axis=1)

# Kaydedip okuma
sf.write("stereo_test.wav", stereo, fs)
data, rate = sf.read("stereo_test.wav")
print(f"Okunan: shape={data.shape}, rate={rate}")
```
